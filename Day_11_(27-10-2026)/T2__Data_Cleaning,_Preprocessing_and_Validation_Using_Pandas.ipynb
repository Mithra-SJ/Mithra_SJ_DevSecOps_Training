```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))

df = pd.read_csv("Online_Retail.csv")

print("Dataset imported successfully!")
print("File name:", filename)

display(df.head())
```

```python
df_original = df.copy()

print("Original dataset backup created!")

print("Dataset shape:", df_original.shape)
display(df_original.head())
```

```python
# First five rows
print("First five rows:")
display(df_original.head())

# First three rows
print("\nFirst three rows:")
display(df_original.head(3))

# Last five rows
print("\nLast five rows:")
display(df_original.tail())

# Dataset dimensions
print("\nDataset shape:", df_original.shape)

# Column names
print("\nColumn names:")
print(df_original.columns.tolist())

# Dataset information
print("\nDataset information:")
df_original.info()

# Descriptive statistics
print("\nDescriptive statistics:")
display(df_original.describe(include="all"))
```

```python
print("Missing values:")
print(df_original.isnull().sum())

print("\nTotal missing values:")
print(df_original.isnull().sum().sum())

print("\nNumber of duplicate records:")
print(df_original.duplicated().sum())
```

```python
df_clean = df_original.copy()

duplicates_before = df_clean.duplicated().sum()

df_clean = df_clean.drop_duplicates().copy()

print("Duplicates removed:", duplicates_before)
print("Shape after removing duplicates:", df_clean.shape)

text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df_clean[col] = df_clean[col].astype("string").str.strip()

if "Country" in df_clean.columns:
    df_clean["Country"] = df_clean["Country"].str.title()

if "Description" in df_clean.columns:
    df_clean["Description"] = df_clean["Description"].str.title()

print("\nText cleaning completed!")
display(df_clean.head())
```

```python
if "InvoiceDate" in df_clean.columns:

    original_dates = df_clean["InvoiceDate"].copy()

    df_clean["InvoiceDate"] = pd.to_datetime(
        df_clean["InvoiceDate"],
        errors="coerce"
    )

    invalid_dates = (
        original_dates.notna() &
        df_clean["InvoiceDate"].isna()
    ).sum()

    print("Date conversion completed!")
    print("InvoiceDate data type:", df_clean["InvoiceDate"].dtype)
    print("Invalid dates:", invalid_dates)

    display(df_clean[["InvoiceDate"]].head())

else:
    print("InvoiceDate column not found!")
```

```python
numeric_columns = [
    "Quantity",
    "UnitPrice",
    "CustomerID"
]

for col in numeric_columns:

    if col in df_clean.columns:

        original_values = df_clean[col].copy()

        df_clean[col] = pd.to_numeric(
            df_clean[col],
            errors="coerce"
        )

        invalid_count = (
            original_values.notna() &
            original_values.astype("string").str.strip().ne("") &
            df_clean[col].isna()
        ).sum()

        print("\nColumn:", col)
        print("Data type:", df_clean[col].dtype)
        print("Invalid values converted to NaN:", invalid_count)

print("\nNumeric conversion completed!")

display(df_clean.head())
```

```python
price = df_clean["UnitPrice"].dropna()

Q1 = price.quantile(0.25)
Q3 = price.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = df_clean[
    (df_clean["UnitPrice"] < lower_bound) |
    (df_clean["UnitPrice"] > upper_bound)
]

print("\nNumber of potential outliers:", len(outliers))

display(outliers.head(10))
```

```python
plt.figure(figsize=(8, 5))

df_clean["UnitPrice"].dropna().plot(kind="box")

plt.title("UnitPrice Outlier Detection")
plt.ylabel("Unit Price")

plt.show()
```

```python
df_clean["Revenue"] = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
)

df_clean["DiscountedPrice"] = (
    df_clean["UnitPrice"] * 0.90
)

df_clean["Year"] = df_clean["InvoiceDate"].dt.year
df_clean["Month"] = df_clean["InvoiceDate"].dt.month

print("New columns created successfully!")

display(
    df_clean[
        [
            "Quantity",
            "UnitPrice",
            "Revenue",
            "DiscountedPrice",
            "Year",
            "Month"
        ]
    ].head(10)
)
```

```python
def categorize_price(price):

    if pd.isna(price):
        return "Unknown"

    elif price < 10:
        return "Low"

    elif price <= 100:
        return "Medium"

    else:
        return "High"


df_clean["PriceCategory"] = df_clean["UnitPrice"].apply(
    categorize_price
)

print("Price categories created!")

display(
    df_clean[
        ["UnitPrice", "PriceCategory"]
    ].head(15)
)

print("\nPrice category counts:")
print(df_clean["PriceCategory"].value_counts())
```

```python
# Check zero or negative quantities
print("Zero or negative quantities:")
display(df_clean[df_clean["Quantity"] <= 0].head(10))

# Check negative prices
print("\nNegative prices:")
display(df_clean[df_clean["UnitPrice"] < 0].head(10))

# Create a separate dataset for positive sales
df_sales = df_clean[
    (df_clean["Quantity"] > 0) &
    (df_clean["UnitPrice"] >= 0) &
    (df_clean["Quantity"].notna()) &
    (df_clean["UnitPrice"].notna())
].copy()

print("\nOriginal cleaned dataset rows:", len(df_clean))
print("Positive sales dataset rows:", len(df_sales))

display(df_sales.head())
```

```python
print("========== FINAL DATA VALIDATION ==========")

print("\nDataset shape:")
print(df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate records:")
print(df_clean.duplicated().sum())

print("\nData types:")
print(df_clean.dtypes)

print("\nDescriptive statistics:")
display(df_clean.describe(include="all"))

new_columns = [
    "Revenue",
    "DiscountedPrice",
    "PriceCategory",
    "Year",
    "Month"
]

print("\nNew columns:")
display(df_clean[new_columns].head(10))
```

```python
expected_revenue = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
)

revenue_correct = np.isclose(
    df_clean["Revenue"].to_numpy(dtype=float, na_value=np.nan),
    expected_revenue.to_numpy(dtype=float, na_value=np.nan),
    equal_nan=True
).all()

print("Revenue calculation correct:", revenue_correct)

expected_discount = df_clean["UnitPrice"] * 0.90

discount_correct = np.isclose(
    df_clean["DiscountedPrice"].to_numpy(dtype=float, na_value=np.nan),
    expected_discount.to_numpy(dtype=float, na_value=np.nan),
    equal_nan=True
).all()

print("Discount calculation correct:", discount_correct)

category_correct = (
    df_clean["PriceCategory"] ==
    df_clean["UnitPrice"].apply(categorize_price)
).all()

print("Price categorization correct:", category_correct)

print("No duplicate records:", df_clean.duplicated().sum() == 0)
```

```python
print("========== ORIGINAL VS CLEANED DATA ==========")

print("Original dataset shape:", df_original.shape)
print("Cleaned dataset shape:", df_clean.shape)

print("\nOriginal duplicate count:",
      df_original.duplicated().sum())

print("Cleaned duplicate count:",
      df_clean.duplicated().sum())

print("\nOriginal missing values:",
      df_original.isnull().sum().sum())

print("Cleaned missing values:",
      df_clean.isnull().sum().sum())

print("\nNew columns added:")

for col in df_clean.columns:
    if col not in df_original.columns:
        print(col)

print("\nOriginal dataset sample:")
display(df_original.head())

print("\nCleaned dataset sample:")
display(df_clean.head())
```

```python
from google.colab import files

# Save cleaned dataset
df_clean.to_csv(
    "Online_Retail_Cleaned.csv",
    index=False
)

print("Cleaned dataset exported successfully!")

# Download the file
files.download("Online_Retail_Cleaned.csv")
```
