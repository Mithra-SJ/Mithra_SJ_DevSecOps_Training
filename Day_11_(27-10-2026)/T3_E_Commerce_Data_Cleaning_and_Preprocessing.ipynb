```python
import pandas as pd
import numpy as np
```

```python
from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))

df = pd.read_csv("Online_Retail.csv")

print("Dataset loaded successfully!")
```

```python
print("First 5 rows:")
display(df.head())

print("Dataset shape:")
print(df.shape)

print("Column names:")
print(df.columns.tolist())

print("Dataset information:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nMissing values percentage:")
print((df.isnull().sum() / len(df)) * 100)
```

```python
df_clean = df.copy()

important_columns = ["InvoiceNo", "Quantity", "UnitPrice"]

df_clean = df_clean.dropna(subset=important_columns)

print("Shape before cleaning:", df.shape)
print("Shape after handling missing important values:", df_clean.shape)

print("\nRemaining missing values:")
print(df_clean.isnull().sum())
```

```python
print("Repeated InvoiceNo values:")
print(df_clean["InvoiceNo"].duplicated().sum())

print("\nExample of repeated invoices:")
display(df_clean[df_clean["InvoiceNo"].duplicated(keep=False)].head(10))
```

```python
print("Exact duplicate rows before removal:")
print(df_clean.duplicated().sum())

df_clean = df_clean.drop_duplicates()

print("Exact duplicate rows after removal:")
print(df_clean.duplicated().sum())
```

```python
df_clean["UnitPrice"] = pd.to_numeric(
    df_clean["UnitPrice"], errors="coerce"
)

valid_prices = df_clean[df_clean["UnitPrice"] >= 0]

highest_price_product = valid_prices.loc[
    valid_prices["UnitPrice"].idxmax()
]

print("Highest-priced product:")
print("Product:", highest_price_product["Description"])
print("Unit Price:", highest_price_product["UnitPrice"])
```

```python
df_clean["Quantity"] = pd.to_numeric(
    df_clean["Quantity"], errors="coerce"
)

df_clean["Revenue"] = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
)

display(df_clean[["InvoiceNo", "Description", "Quantity", "UnitPrice", "Revenue"]].head())
```

```python
country_revenue = df_clean.groupby("Country")["Revenue"].sum()

print("Revenue by country:")
print(country_revenue.sort_values(ascending=False))
```

```python
highest_revenue_country = country_revenue.idxmax()
highest_revenue = country_revenue.max()

print("Highest revenue country:", highest_revenue_country)
print("Total revenue:", highest_revenue)
```

```python
product_revenue = (
    df_clean.groupby("Description")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 products by revenue:")
display(product_revenue.head(10))
```

```python
print("Rows with negative unit prices:")
display(df_clean[df_clean["UnitPrice"] < 0].head())

print("Rows with zero unit prices:")
display(df_clean[df_clean["UnitPrice"] == 0].head())

print("Rows with zero or negative quantities:")
display(df_clean[df_clean["Quantity"] <= 0].head())
```

```python
print("Unique country names before standardization:")
print(df_clean["Country"].unique())
```

```python
df_clean["Country"] = df_clean["Country"].astype("string").str.strip().str.title()

print("Country names after standardization:")
print(df_clean["Country"].unique())
```

```python
print("Final dataset shape:")
print(df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())

print("\nFinal dataset preview:")
display(df_clean.head())
```

```python
df_clean.to_csv("Online_Retail_Business_Analysis_Cleaned.csv", index=False)

from google.colab import files
files.download("Online_Retail_Business_Analysis_Cleaned.csv")
```
