```python
filename = next(iter(uploaded))
df = pd.read_csv("Online_Retail.csv")

display(df.head())

print(df.columns.tolist())

print("Rows and columns:", df.shape)
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]

display(df.head())
```

```python
print("Total quantity:", df["Quantity"].sum())
print("Average quantity:", df["Quantity"].mean())
print("Minimum quantity:", df["Quantity"].min())
print("Maximum quantity:", df["Quantity"].max())
print("Number of quantity records:", df["Quantity"].count())

print("\nTotal revenue:", df["Revenue"].sum())
print("Average revenue per row:", df["Revenue"].mean())
print("Minimum revenue:", df["Revenue"].min())
print("Maximum revenue:", df["Revenue"].max())

print("\nAggregate statistics for numeric columns:")
display(df.agg({
    "Quantity": ["sum", "mean", "min", "max", "count"],
    "UnitPrice": ["sum", "mean", "min", "max", "count"],
    "Revenue": ["sum", "mean", "min", "max", "count"]
}))
```

```python
# Rows where revenue is greater than 2000
high_revenue = df[df["Revenue"] > 2000]

print("Rows with revenue greater than 2000:")
display(high_revenue.head(10))

# Rows where quantity is greater than 5
high_quantity = df[df["Quantity"] > 5]

print("\nRows where quantity is greater than 5:")
display(high_quantity.head(10))

# Compare rows with positive and negative quantities
print("\nPositive quantity rows:", (df["Quantity"] > 0).sum())
print("Negative quantity rows:", (df["Quantity"] < 0).sum())

# Compare rows with positive and negative prices
print("Positive price rows:", (df["UnitPrice"] > 0).sum())
print("Zero or negative price rows:", (df["UnitPrice"] <= 0).sum())
```

```python
# Quantity greater than 2 AND unit price greater than 100
filtered_data = df[
    (df["Quantity"] > 2) &
    (df["UnitPrice"] > 100)
]

print("Records matching both conditions:")
display(filtered_data.head(10))

# Filter by country
country_data = df[df["Country"] == "United Kingdom"]

print("\nUnited Kingdom records:")
display(country_data.head(10))

# Filter using a country and revenue condition
country_revenue = df[
    (df["Country"] == "United Kingdom") &
    (df["Revenue"] > 2000)
]

print("\nUK records with revenue greater than 2000:")
display(country_revenue.head(10))
```

```python
conditions = [
    df["Revenue"] < 100,
    (df["Revenue"] >= 100) & (df["Revenue"] <= 1000),
    df["Revenue"] > 1000
]

categories = ["Low", "Medium", "High"]

df["RevenueCategory"] = np.select(
    conditions,
    categories,
    default="Unknown"
)

print("Revenue categories:")
display(df[["Quantity", "UnitPrice", "Revenue", "RevenueCategory"]].head(10))

print("\nCount of records in each category:")
print(df["RevenueCategory"].value_counts())
```

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

print("Date column converted successfully!")

print("InvoiceDate data type:", df["InvoiceDate"].dtype)

display(df[["InvoiceDate"]].head())

print("Invalid or missing dates:", df["InvoiceDate"].isna().sum())
```

```python
df["Year"] = df["InvoiceDate"].dt.year
df["Month"] = df["InvoiceDate"].dt.month
df["Day"] = df["InvoiceDate"].dt.day
df["MonthName"] = df["InvoiceDate"].dt.month_name()
df["DayOfWeek"] = df["InvoiceDate"].dt.day_name()

print("New date columns created:")
display(
    df[["InvoiceDate", "Year", "Month", "MonthName", "Day", "DayOfWeek"]].head(10)
)
```

```python
revenue_by_country = (
    df.groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 countries by revenue:")
display(revenue_by_country.head(10))

if "Description" in df.columns:
    revenue_by_product = (
        df.groupby("Description")["Revenue"]
        .sum()
        .sort_values(ascending=False)
    )

    print("\nTop 10 products by revenue:")
    display(revenue_by_product.head(10))

monthly_revenue = (
    df.dropna(subset=["InvoiceDate"])
    .groupby(df["InvoiceDate"].dt.to_period("M"))["Revenue"]
    .sum()
)

monthly_revenue.index = monthly_revenue.index.astype(str)

print("\nMonthly revenue:")
display(monthly_revenue)
```

```python
top_countries = revenue_by_country.head(10).sort_values()

top_countries.plot(kind="barh", figsize=(10, 6))
plt.title("Top 10 Countries by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

```python
monthly_revenue.plot(kind="line", marker="o", figsize=(12, 5))
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
df["RevenueCategory"].value_counts().reindex(
    ["Low", "Medium", "High", "Unknown"],
    fill_value=0
).plot(kind="bar", figsize=(8, 5))

plt.title("Number of Records by Revenue Category")
plt.xlabel("Revenue Category")
plt.ylabel("Number of Records")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
print("========== FINAL DATA ANALYSIS SUMMARY ==========")

print("Total records:", len(df))
print("Total columns:", len(df.columns))
print("Total quantity:", df["Quantity"].sum())
print("Average quantity:", round(df["Quantity"].mean(), 2))
print("Total row-level revenue:", round(df["Revenue"].sum(), 2))
print("Average row-level revenue:", round(df["Revenue"].mean(), 2))

print("\nNumber of countries:", df["Country"].nunique())
print("Missing values:", df.isnull().sum().sum())

print("\nCountry with the highest total revenue:")
print(revenue_by_country.idxmax(), "-", round(revenue_by_country.max(), 2))

if "Description" in df.columns:
    print("\nProduct with the highest total revenue:")
    print(revenue_by_product.idxmax(), "-", round(revenue_by_product.max(), 2))

print("\nRevenue category counts:")
print(df["RevenueCategory"].value_counts())

print("\nAnalysis completed successfully!")
```
